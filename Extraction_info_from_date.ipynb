```python
import pandas as pd
import numpy as np
```

```python
startDate = '2017-07-06'  # YYYY-MM-DD
endDate = '2017-07-18'
Dates = pd.DataFrame(pd.date_range(startDate, endDate), columns = ['Dates'])
```

```python
Dates
```

```python
Dates['day'] = Dates['Dates'].map(lambda x:x.day)
Dates['month'] = Dates['Dates'].map(lambda x:x.month)
Dates['day_of_week'] = Dates['Dates'].map(lambda x:x.day_name())
# Dates['is_weekend'] = Dates['Dates'].map(lambda x:np.is_busday(x, weekmask = '1111100'))
# Dates['is_weekend'] = Dates['Dates'].map(lambda x: not np.is_busday(x, weekmask='1111100'))
Dates['is_weekend'] = Dates['Dates'].map(lambda x: x.weekday() >= 5)
```
